Handle missing data:

- For continuous attributes (e.g., age), replace missing values with the mean.
- For categorical attributes (e.g., smoker), replace missing values with the most frequent value.
- Update the data types of the respective columns.
- Verify the update using `df.info()`.


````python
# 1. 处理 age 列
# 计算均值
mean_age = df['age'].astype('float').mean()
df["age"] = df["age"].replace(np.nan, mean_age)

# 2. 处理 smoker 列
# 计算众数
if_smoker = df['smoker'].value_counts().idxmax()
df["smoker"] = df["smoker"].replace(np.nan, if_smoker)

# 3. 类型转换
df["age"] = df["age"].astype("int") 
df['smoker'] = df['smoker'].astype('int')
````

# Task 4 : Model Development

Fit a linear regression model that may be used to predict the `charges` value, just by using the `smoker` attribute of the dataset. Print the $ R^2 $ score of this model.


````python
X = df[['smoker']]
Y = df['charges']
lm = LinearRegression()
lm.fit(X,Y)
score = lm.score(X, Y)
print(f"The R-squared score on the dataset is: {score:.4f}")
````

Create a training pipeline that uses `StandardScaler()`, `PolynomialFeatures()` and `LinearRegression()` to create a model that can predict the `charges` value using all the other attributes of the dataset. There should be even further improvement in the performance.


````python
# 构建 Pipeline
steps =[('scale',StandardScaler()), ('polynomial', PolynomialFeatures(include_bias=False)), ('model', LinearRegression())]
pipe=Pipeline(steps)
Z = Z.astype(float)
pipe.fit(Z,Y)
y_pipe=pipe.predict(Z)
score = r2_score(Y, y_pipe)
print(f"The R-squared score on the test set (with Pipeline) is: {score:.4f}")
````

Initialize a Ridge regressor that used hyperparameter $ \alpha = 0.1 $. Fit the model using training data data subset. Print the $ R^2 $ score for the testing data.

Apply polynomial transformation to the training parameters with degree=2. Use this transformed feature set to fit the same regression model, as above, using the training subset. Print the $ R^2 $ score for the testing subset.


````python
# 1. 初始化多项式特征转换器
pr = PolynomialFeatures(degree=2)
# 2. 转换训练集（拟合+转换）
Z_train_pr = pr.fit_transform(Z_train)
# 3. 转换测试集（仅转换）
Z_test_pr = pr.fit_transform(Z_test)
# 4. 训练岭回归模型
RidgeModel.fit(Z_train_pr, Y_train)
# 5. 模型预测
y_hat = RidgeModel.predict(Z_test_pr)
# 6. 评估模型性能
print(r2_score(Y_test,y_hat))
````
